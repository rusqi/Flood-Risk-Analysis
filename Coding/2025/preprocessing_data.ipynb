```python
import pandas as pd

# 1. Membaca data dari file Excel
file_path = "Data Curah Hujan Harian 2022-2024.xlsx"  # Path ke file Excel
data = pd.read_excel(file_path, skiprows=8)  # Membaca data dari file Excel

# 2. Menetapkan nama kolom secara manual
data.columns = ['Tanggal', 'Curah Hujan', 'Kolom_Tidak_Terpakai', 'Keterangan'] # Nama kolom ditetapkan secara manual
# 3. Menghapus kolom yang tidak relevan
data = data.drop(columns=['Kolom_Tidak_Terpakai', 'Keterangan'])  # Menghapus kolom yang tidak digunakan
# 3. Menampilkan beberapa baris pertama untuk memeriksa data
# 4. Mengonversi 'Curah Hujan' menjadi numerik
data['Curah Hujan'] = pd.to_numeric(data['Curah Hujan'], errors='coerce')  # Mengonversi nilai curah hujan menjadi numerik

# 5. Mengonversi 'Tanggal' menjadi format datetime
data['Tanggal'] = pd.to_datetime(data['Tanggal'], errors='coerce')  # Mengonversi kolom Tanggal menjadi format datetime

# 6. Menampilkan data setelah konversi untuk memastikan perubahan
print(data)
```

```python
# Fungsi untuk menerapkan Mean Imputation pada missing values
def mean_imputation(row, data):
    # Mendapatkan index baris saat ini
    idx = row.name
    # Cek apakah nilai saat ini adalah missing
    if pd.isna(row['Curah Hujan']):
        # Mengambil nilai curah hujan dua hari sebelumnya dan dua hari setelahnya
        prev_2 = data.iloc[idx - 2]['Curah Hujan'] if idx - 2 >= 0 else None
        prev_1 = data.iloc[idx - 1]['Curah Hujan'] if idx - 1 >= 0 else None
        next_1 = data.iloc[idx + 1]['Curah Hujan'] if idx + 1 < len(data) else None
        next_2 = data.iloc[idx + 2]['Curah Hujan'] if idx + 2 < len(data) else None

        # Mengambil nilai non-NaN untuk menghitung rata-rata
        values = [value for value in [prev_2, prev_1, next_1, next_2] if pd.notna(value)]
        
        if values:
            return round(sum(values) / len(values), 1)
    return row['Curah Hujan']  # Jika tidak missing, return nilai asli

# 6. Menerapkan mean imputation pada kolom 'Curah Hujan'
data['Curah Hujan'] = data.apply(mean_imputation, axis=1, data=data)

# 7. Menyimpan hasil data yang telah dibersihkan ke file Excel baru
output_file_path = 'Data Curah Hujan Harian Cleaned.xlsx'
data.to_excel(output_file_path, index=False)

# Menampilkan path file hasil output
output_file_path
```
